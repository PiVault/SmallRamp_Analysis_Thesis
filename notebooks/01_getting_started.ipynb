# Getting Started with balancepy

This is a template notebook to get started with the data analysis project.

```python
import numpy as np
import pandas as pd
import plotly.express as px
import balancepy as bp
from scipy import signal
```

## Import balancepy modules

```python
# Check balancepy is installed
print('balancepy imported successfully!')
print(f'balancepy location: {bp.__file__}')
```

## Load and Process Data

Example: Load data and compute center of mass

```python
# Generate example data
t = np.linspace(0, 10, 1000)
shoulder_motion = 0.05 * np.sin(2*np.pi*0.5*t)
hip_motion = 0.03 * np.sin(2*np.pi*0.3*t)

# Compute center of mass
com = bp.biomechanics.get_com(
    shoulder_t=shoulder_motion,
    shoulder_marker_height=1.5,
    hip_t=hip_motion,
    hip_marker_height=0.9,
    height_m=1.75,
    rotation=True
)

print(f'COM angle range: {np.min(com):.2f} to {np.max(com):.2f} degrees')
```

## Generate Test Stimulus

```python
# Create a sine wave stimulus
stimulus = bp.make_stimulus.make_sine(
    frequency_Hz=1,
    ncyc=5,
    ampl=0.1,
    samplingrate_Hz=100
)

print(f'Stimulus shape: {stimulus.shape}')
print(f'Stimulus amplitude: {np.max(np.abs(stimulus)):.4f}')
```

## Frequency Domain Analysis

```python
# Compute frequency spectrum
Sx, Sxx, freq = bp.frequency.spectrum(stimulus, samplingrate_Hz=100)

# Plot power spectrum
fig = px.line(x=freq, y=np.abs(Sx), 
              labels={'x': 'Frequency (Hz)', 'y': 'Amplitude'},
              title='Power Spectrum')
fig.show()
```
